##**OUTILERS DETECTION**

```python
a=df.select_dtypes(include=np.number)
```

```python
sns.boxplot([df["Survived"],df["Pclass"],df["SibSp"],df["Parch"]])
```

##**IQR**

```python
for i in a:
    q1=a[i].quantile(0.25)
    q2=a[i].quantile(0.50)
    q3=a[i].quantile(0.75)
```

a=dataframe columns

loop runs through each column one by one

i reperesents each colum name

q1 lower quartile value

q2 is median quartile value

q3 upper quartile value

```python
IQR=q3-q1
```

  IQR inter quartile range

  measures spread of middle 50%data



```python
Uw=q3+1.5*IQR
Lw=q1-1.5*IQR
```

uw upper whisker\upper limit

values greater than this are outliers

LW lower whisker\lower limit

value smaller than this are outliers

```python
outliers=((a[i]>Uw)|(a[i]<Lw))
```

greater than upper limit or smaller than lower limit

means or condition

returns:

true: outliers

false : normal value

```python
print(outliers.sum())
```

```python
a.loc[(a[i]>Uw)|(a[i]<Lw),i]=q2
```

finds outlier rows

replace them with q2

q2=median value

median is less affected by outliers

```python
print(outliers.sum())
```

##**Binning**

```python
bins=[0,18,40,70]
labels=["Child","adult","old"]
df["Agegroup"]=pd.cut(df["Age"],bins=bins,labels=labels)
```

```python
c=df.groupby("Agegroup")["Survived"].mean()*100
```
